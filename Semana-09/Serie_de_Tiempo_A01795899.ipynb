### Preparación de los datos

```python
file_name = 'monthly_champagne_sales.csv'
data = pd.read_csv(file_name)
data.head()
```

```python
data.info() # Observamos que la fecha es una variable de tipo objeto o categórica.
```

```python
# Renombramos de manera estándar las columnas de fechas y valores de
# la serie de tiempo, donde además la fehhca la transformamos a tipo datetime:
data.columns = ['ds', 'y']
data['ds'] = pd.to_datetime(data['ds'])
data.head()
```

```python
# Hacemos una copia del data frame 'data' con indices temporales:
data_temp = data.copy()
data_temp.set_index('ds', inplace=True)
data_temp.head()
```

```python
print(data.shape)
print(data_temp.shape)
```

```python
# Ajustamos la frecuencia de los índices, para que el modelo tenga información del período en que
# se encuentran los datos:
# Especifiquamos entonces que nuestros datos están registrados con una frecuencia mensual:
data_temp.index.freq = 'MS'  # MS-Month-start-frequency o M=ME-Month-end-frequency
data_temp.index
```

```python
# Particionamos en entrenamiento y prueba como DataFrames de Pandas:
train = data_temp.iloc[0:-12,:]
test = data_temp.iloc[-12:,:]   # Seleccionamos los datos del último año para predicción de Test.
```
